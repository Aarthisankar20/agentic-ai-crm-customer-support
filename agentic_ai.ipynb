```python
# ============================================
# CELL 24 — LOAD EMBEDDING MODEL
# ============================================

from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

print("✅ Embedding model loaded")
```

```python
# ============================================
# CELL 25 — DOCUMENT CHUNKING
# ============================================

chunks = []

for filename in documents:

    path = os.path.join(documents_path, filename)

    with open(path, "r", encoding="utf-8") as f:
        text = f.read()

    # Split by paragraphs
    paragraphs = [
        p.strip()
        for p in text.split("\n\n")
        if p.strip()
    ]

    for i, paragraph in enumerate(paragraphs):

        chunks.append({
            "chunk_id": f"{filename}_{i}",
            "source": filename,
            "text": paragraph
        })

chunks_df = pd.DataFrame(chunks)

print("✅ Document chunks created")
print("Number of chunks:", len(chunks_df))

display(chunks_df)
```

```python
# ============================================
# CELL 26 — CREATE DOCUMENT EMBEDDINGS
# ============================================

texts = chunks_df["text"].tolist()

embeddings = embedding_model.encode(
    texts,
    convert_to_numpy=True,
    show_progress_bar=True
)

print("✅ Embeddings created")
print("Embedding shape:", embeddings.shape)
```

```python
# ============================================
# CELL 27 — CREATE VECTOR DATABASE
# ============================================

import faiss
import numpy as np

dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(
    np.array(embeddings).astype("float32")
)

print("✅ FAISS vector database created")
print("Vectors stored:", index.ntotal)
```

```python
# ============================================
# CELL 28 — RAG RETRIEVER
# ============================================

def retrieve_documents(query, top_k=3):

    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True
    )

    distances, indices = index.search(
        np.array(query_embedding).astype("float32"),
        top_k
    )

    results = []

    for distance, idx in zip(distances[0], indices[0]):

        results.append({
            "source": chunks_df.iloc[idx]["source"],
            "text": chunks_df.iloc[idx]["text"],
            "distance": float(distance)
        })

    return results
```

```python
# ============================================
# CELL 30 — CRM DATABASE TOOL
# ============================================

def get_customer_orders(customer_id):
    """
    Retrieve all orders belonging to a customer.
    """

    query = """
    SELECT
        o.order_id,
        o.customer_id,
        p.product_name,
        p.category,
        o.order_date,
        o.quantity,
        o.total_amount,
        o.payment_status,
        o.delivery_status
    FROM orders o
    LEFT JOIN products p
        ON o.product_id = p.product_id
    WHERE o.customer_id = ?
    ORDER BY o.order_date DESC
    """

    result = pd.read_sql_query(
        query,
        conn,
        params=(customer_id,)
    )

    return result
```

```python
# ============================================
# CELL 33 — KNOWLEDGE BASE TOOL
# ============================================

def search_knowledge_base(query, top_k=3):

    results = retrieve_documents(
        query,
        top_k=top_k
    )

    return results
```

```python
# ============================================
# CELL 35 — REFUND INVESTIGATION WORKFLOW
# ============================================

def investigate_refund(customer_id):

    print("🤖 Starting refund investigation...")

    # ----------------------------------------
    # STEP 1 — CRM
    # ----------------------------------------
    print("\n🔍 Step 1: Checking customer orders...")

    orders = get_customer_orders(customer_id)

    if orders.empty:
        return {
            "status": "error",
            "message": "Customer or orders not found."
        }

    print(f"Found {len(orders)} orders.")

    # ----------------------------------------
    # STEP 2 — Find delayed orders
    # ----------------------------------------
    print("\n📦 Step 2: Checking delayed orders...")

    delayed_orders = orders[
        orders["delivery_status"].str.lower() == "delayed"
    ]

    print(
        f"Delayed orders found: "
        f"{len(delayed_orders)}"
    )

    # ----------------------------------------
    # STEP 3 — RAG
    # ----------------------------------------
    print("\n📚 Step 3: Checking refund policy...")

    policy_results = search_knowledge_base(
        "refund policy for delayed orders",
        top_k=3
    )

    # ----------------------------------------
    # STEP 4 — Return evidence
    # ----------------------------------------
    return {
        "customer_id": customer_id,
        "orders": orders,
        "delayed_orders": delayed_orders,
        "policy_results": policy_results
    }
```

```python
# ============================================
# CELL 36 — RUN FIRST AI WORKFLOW
# ============================================

investigation = investigate_refund(
    customer_id
)

print("\n✅ Investigation completed")
```

```python
# ============================================
# CELL 42 — GEMINI LLM FUNCTION
# ============================================

import requests
import json

def ask_gemini(prompt):

    url = (
        "https://generativelanguage.googleapis.com/v1beta/"
        "models/gemini-3.6-flash:generateContent"
        f"?key={GEMINI_API_KEY}"
    )

    payload = {
        "contents": [
            {
                "parts": [
                    {
                        "text": prompt
                    }
                ]
            }
        ]
    }

    response = requests.post(
        url,
        headers={
            "Content-Type": "application/json"
        },
        json=payload,
        timeout=60
    )

    if response.status_code != 200:
        raise Exception(
            f"Gemini API Error {response.status_code}: "
            f"{response.text}"
        )

    data = response.json()

    return data["candidates"][0]["content"]["parts"][0]["text"]
```

```python
# ============================================
#  AGENT TOOL EXECUTOR
# ============================================

def execute_agent(question):

    print("🤖 Supervisor Agent is analyzing the request...\n")

    decision = supervisor_agent(question)

    print("SUPERVISOR DECISION")
    print("=" * 60)
    print(decision)

    # ----------------------------------------
    # Determine requested actions
    # ----------------------------------------

    decision_upper = decision.upper()

    crm_needed = "CRM" in decision_upper
    rag_needed = "RAG" in decision_upper

    # ----------------------------------------
    # Extract customer ID
    # ----------------------------------------

    import re

    customer_match = re.search(
        r"CUST\d+",
        decision.upper()
    )

    customer_id = (
        customer_match.group(0)
        if customer_match
        else None
    )

    # ----------------------------------------
    # CRM TOOL
    # ----------------------------------------

    crm_result = None

    if crm_needed and customer_id:

        print("\n🔍 Calling CRM Tool...")

        crm_result = get_customer_orders(
            customer_id
        )

        print(
            f"✅ CRM returned "
            f"{len(crm_result)} orders"
        )

    # ----------------------------------------
    # RAG TOOL
    # ----------------------------------------

    rag_result = None

    if rag_needed:

        print("\n📚 Calling RAG Tool...")

        rag_result = search_knowledge_base(
            question,
            top_k=3
        )

        print(
            f"✅ RAG returned "
            f"{len(rag_result)} documents"
        )

    return {
        "decision": decision,
        "customer_id": customer_id,
        "crm_result": crm_result,
        "rag_result": rag_result
    }
```

```python

# CELL 68 — LOAD DOCUMENTS

documents_data = []

for filename in os.listdir(DOCUMENT_PATH):

    file_path = os.path.join(
        DOCUMENT_PATH,
        filename
    )

    with open(
        file_path,
        "r",
        encoding="utf-8"
    ) as f:

        text = f.read()

    documents_data.append({
        "source": filename,
        "text": text
    })


print(
    f"✅ Loaded {len(documents_data)} documents"
)

for doc in documents_data:
    print(
        f"\n📄 {doc['source']} "
        f"→ {len(doc['text'])} charaters"
    )
```

```python

# CELL 69 — DOCUMENT CHUNKING

chunks = []

chunk_size = 800
overlap = 150

for doc in documents_data:

    text = doc["text"]

    start = 0

    while start < len(text):

        end = start + chunk_size

        chunk_text = text[start:end]

        chunks.append({
            "source": doc["source"],
            "text": chunk_text
        })

        start = end - overlap


print("✅ Total chunks:", len(chunks))

print("\nExample chunk:")
print("=" * 60)
print(chunks[0]["text"])
```

```python
# ============================================
#  CREATE DOCUMENT EMBEDDINGS
# ============================================

from sentence_transformers import SentenceTransformer
import numpy as np

print("📥 Loading embedding model...")

embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

print("✅ Embedding model loaded")

texts = [
    chunk["text"]
    for chunk in chunks
]

embeddings = embedding_model.encode(
    texts,
    convert_to_numpy=True,
    show_progress_bar=True
)

print("\n✅ Embeddings created")
print("Number of chunks:", len(embeddings))
print("Embedding dimension:", embeddings.shape[1])
```

```python
# ============================================
# CREATE FAISS VECTOR DATABASE
# ============================================

import faiss

dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(
    embeddings.astype("float32")
)

print("✅ FAISS vector database created")
print("Vectors stored:", index.ntotal)
```

```python
# ============================================
# SAVE RAG INDEX
# ============================================

RAG_PATH = (
    "/content/drive/MyDrive/Agentic_AI_CRM/rag"
)

import os
import pickle

os.makedirs(
    RAG_PATH,
    exist_ok=True
)

# Save FAISS index
faiss.write_index(
    index,
    os.path.join(
        RAG_PATH,
        "knowledge_base.index"
    )
)

# Save chunks
with open(
    os.path.join(
        RAG_PATH,
        "chunks.pkl"
    ),
    "wb"
) as f:

    pickle.dump(
        chunks,
        f
    )

print("✅ FAISS index saved")
print("✅ Document chunks saved")
print("📁 Location:", RAG_PATH)
```

```python
# ============================================
# RAG RETRIEVER
# ============================================

def search_knowledge_base(
    query,
    top_k=3
):

    # Convert query to embedding
    query_embedding = embedding_model.encode(
        [query],
        convert_to_numpy=True
    ).astype("float32")

    # Search FAISS
    distances, indices = index.search(
        query_embedding,
        top_k
    )

    results = []

    for distance, idx in zip(
        distances[0],
        indices[0]
    ):

        if idx == -1:
            continue

        results.append({
            "source": chunks[idx]["source"],
            "text": chunks[idx]["text"],
            "distance": float(distance)
        })

    return results
```

```python
# ============================================
# PARSE AGENT DECISION
# ============================================

import re

def parse_decision(text):

    action = re.search(
        r"ACTION:\s*(.*)",
        text
    )

    customer = re.search(
        r"CUSTOMER_ID:\s*(.*)",
        text
    )

    rag = re.search(
        r"RAG_QUERY:\s*(.*)",
        text
    )

    return {
        "action": action.group(1).strip()
        if action else "NONE",

        "customer_id": customer.group(1).strip()
        if customer else None,

        "rag_query": rag.group(1).strip()
        if rag else None,

        "raw": text
    }

print("✅ Decision parser ready")
```

```python
# ============================================
# AGENT EXECUTOR
# ============================================

def execute_agent(question):

    decision_text = supervisor_agent(question)

    decision = parse_decision(decision_text)

    action = decision["action"].upper()

    customer_id = decision["customer_id"]

    rag_query = decision["rag_query"]

    result = {
        "decision": decision,
        "profile": None,
        "tickets": None,
        "messages": None,
        "rag": None
    }

    print("🤖 Supervisor Decision")
    print(decision_text)

    if action in ["CRM", "BOTH"]:

        print("\n🔎 Calling CRM tools...")

        if customer_id and customer_id != "NONE":

            result["profile"] = get_customer_profile(customer_id)

            result["tickets"] = get_customer_ticket(customer_id)

            result["messages"] = get_customer_messages(customer_id)

            print("✅ CRM retrieved")

        else:

            print("⚠️ No valid customer ID")

    if action in ["RAG", "BOTH"]:

        print("\n📚 Searching knowledge base...")

        query = rag_query if rag_query != "NONE" else question

        result["rag"] = search_knowledge_base(
            query,
            top_k=3
        )

        print("✅ Policy retrieved")

    return result
```

```python
# ============================================
# FINAL RESPONSE AGENT
# ============================================

def final_response(question, result):

    evidence = ""

    if result["profile"] is not None:
        evidence += "\nCUSTOMER PROFILE:\n"
        evidence += result["profile"].to_string(index=False)

    if result["tickets"] is not None:
        evidence += "\n\nSUPPORT TICKETS:\n"
        evidence += result["tickets"].to_string(index=False)

    if result["messages"] is not None:
        evidence += "\n\nCONVERSATION HISTORY:\n"
        evidence += result["messages"].head(20).to_string(index=False)

    if result["rag"] is not None:

        evidence += "\n\nCOMPANY POLICIES:\n"

        for item in result["rag"]:

            evidence += f"\nSource: {item['source']}\n"
            evidence += item["text"] + "\n"

    prompt = f"""
You are the Final Response Agent of an Agentic CRM.

Answer the user's question ONLY using the evidence below.

USER QUESTION:
{question}

EVIDENCE:
{evidence}

Rules:
- Never invent facts.
- Clearly explain findings.
- Mention relevant policy.
- If policy requires human review, clearly state it.
- Be professional and concise.
"""

    return ask_gemini(prompt)
```

```python
# ============================================
# AGENT STATE
# ============================================

class AgentState(TypedDict, total=False):

    question: str

    action: str

    customer_id: Optional[str]

    rag_query: Optional[str]

    crm_data: Any

    rag_data: Any

    analysis: str

    final_response: str

    escalation_required: bool


print("✅ Agent state created")
```

```python
# ============================================
#  CRM AGENT
# ============================================

def crm_agent(state: AgentState):

    customer_id = state.get("customer_id")

    print("\n🔵 CRM AGENT")

    if not customer_id or customer_id == "NONE":

        print("⚠️ No customer ID provided")

        return {
            "crm_data": None
        }

    profile = get_customer_profile(
        customer_id
    )

    tickets = get_customer_ticket(
        customer_id
    )

    messages = get_customer_messages(
        customer_id
    )

    crm_data = {
        "profile": profile,
        "tickets": tickets,
        "messages": messages
    }

    print(
        f"✅ Retrieved CRM data for {customer_id}"
    )

    return {
        "crm_data": crm_data
    }
```

```python

# ============================================
# RAG AGENT
# ============================================

def rag_agent(state: AgentState):

    question = state.get(
        "rag_query"
    )

    if not question or question == "NONE":

        question = state["question"]

    print("\n🟢 RAG AGENT")

    results = search_knowledge_base(
        question,
        top_k=3
    )

    print(
        f"✅ Retrieved {len(results)} policy chunks"
    )

    return {
        "rag_data": results
    }
```

```python
# ============================================
# SUPPORT AGENT
# ============================================

def support_agent(state: AgentState):

    print("\n🟡 SUPPORT AGENT")

    question = state["question"]

    crm_data = state.get(
        "crm_data"
    )

    rag_data = state.get(
        "rag_data"
    )

    evidence = ""

    # CRM evidence
    if crm_data:

        evidence += "\nCUSTOMER DATA:\n"

        evidence += (
            crm_data["profile"]
            .to_string(index=False)
        )

        evidence += "\n\nTICKETS:\n"

        evidence += (
            crm_data["tickets"]
            .to_string(index=False)
        )

        evidence += "\n\nMESSAGES:\n"

        evidence += (
            crm_data["messages"]
            .head(20)
            .to_string(index=False)
        )

    # RAG evidence
    if rag_data:

        evidence += "\n\nPOLICY EVIDENCE:\n"

        for item in rag_data:

            evidence += (
                f"\nSource: {item['source']}\n"
            )

            evidence += item["text"]

    prompt = f"""
You are the Support Analysis Agent.

USER QUESTION:
{question}

VERIFIED EVIDENCE:
{evidence}

Analyze the case.

Determine:
1. What is the customer's issue?
2. What does the CRM show?
3. What policy applies?
4. What should the support team do?
5. Does this require human escalation?

Never invent information.
"""

    analysis = ask_gemini(prompt)

    print("✅ Support analysis completed")

    return {
        "analysis": analysis
    }
```

```python
# ============================================
# ESCALATION AGENT
# ============================================

def escalation_agent(state: AgentState):

    print("\n🔴 ESCALATION AGENT")

    analysis = state.get(
        "analysis",
        ""
    )

    prompt = f"""
You are the Escalation Decision Agent.

Review this support analysis:

{analysis}

Determine whether human intervention is required.

Return ONLY:

ESCALATE: YES

or

ESCALATE: NO

Escalate when:
- Financial approval is required
- Refund approval is required
- Fraud is suspected
- Safety concerns exist
- Customer requests a human
- The issue remains unresolved
- The AI lacks sufficient evidence
"""

    decision = ask_gemini(prompt)

    escalate = (
        "ESCALATE: YES"
        in decision.upper()
    )

    print(
        "🚨 Escalation:",
        "YES" if escalate else "NO"
    )

    return {
        "escalation_required": escalate
    }
```

```python
# ============================================
# CELL 92 — FINAL RESPONSE AGENT
# ============================================

def final_agent(state: AgentState):

    print("\n🟣 FINAL RESPONSE AGENT")

    question = state["question"]

    analysis = state.get(
        "analysis",
        ""
    )

    escalation = state.get(
        "escalation_required",
        False
    )

    escalation_text = (
        "Human support escalation is recommended."
        if escalation
        else
        "Human escalation is not currently required."
    )

    prompt = f"""
You are the final customer-support response agent.

USER QUESTION:
{question}

SUPPORT ANALYSIS:
{analysis}

ESCALATION STATUS:
{escalation_text}

Create a professional customer-support response.

Rules:
- Use only verified information.
- Do not invent customer/order details.
- Do not claim a refund has been approved unless explicitly confirmed.
- Clearly explain the next step.
- If escalation is required, tell the customer.
"""

    response = ask_gemini(prompt)

    print("✅ Final response generated")

    return {
        "final_response": response
    }
```

```python
# ============================================
#  BUILD AGENT GRAPH
# ============================================

graph = StateGraph(AgentState)

# Add agents
graph.add_node(
    "supervisor",
    supervisor_agent
)

graph.add_node(
    "crm",
    crm_agent
)

graph.add_node(
    "rag",
    rag_agent
)

graph.add_node(
    "support",
    support_agent
)

graph.add_node(
    "escalation",
    escalation_agent
)

graph.add_node(
    "final",
    final_agent
)
```

```python
# ============================================
# SUPERVISOR NODE
# ============================================

def supervisor_node(state: AgentState):

    question = state["question"]

    print("\n🤖 SUPERVISOR AGENT")

    response = supervisor_agent(
        question
    )

    decision = parse_decision(
        response
    )

    print(response)

    return {
        "action": decision["action"],
        "customer_id": decision["customer_id"],
        "rag_query": decision["rag_query"]
    }
```

```python
graph = StateGraph(AgentState)

graph.add_node(
    "supervisor",
    supervisor_node
)

graph.add_node(
    "crm",
    crm_agent
)

graph.add_node(
    "rag",
    rag_agent
)

graph.add_node(
    "support",
    support_agent
)

graph.add_node(
    "escalation",
    escalation_agent
)

graph.add_node(
    "final",
    final_agent
)

print("✅ Nodes registered")
```

```python
# ============================================
#  AGENT ROUTING
# ============================================

def route_after_supervisor(state):

    action = state.get(
        "action",
        "NONE"
    ).upper()

    if action == "CRM":
        return "crm"

    if action == "RAG":
        return "rag"

    if action == "BOTH":
        return "crm"

    return "support"
```

```python
graph.add_edge(
    START,
    "supervisor"
)

graph.add_conditional_edges(
    "supervisor",
    route_after_supervisor,
    {
        "crm": "crm",
        "rag": "rag",
        "support": "support"
    }
)

graph.add_edge(
    "crm",
    "rag"
)

graph.add_edge(
    "rag",
    "support"
)

graph.add_edge(
    "support",
    "escalation"
)

graph.add_edge(
    "escalation",
    "final"
)

graph.add_edge(
    "final",
    END
)

print("✅ Agent workflow connected")
```

```python
# ============================================
# CELL 104 — CREATE CRM BACKEND
# ============================================

import os

BACKEND_PATH = "/content/crm_backend.py"

backend_code = r'''
import sqlite3
import pandas as pd

DB_PATH = "/content/drive/MyDrive/Agentic_AI_CRM/database/crm.db"


def get_connection():
    return sqlite3.connect(
        DB_PATH,
        check_same_thread=False
    )


def get_database_tables():

    conn = get_connection()

    query = """
    SELECT name
    FROM sqlite_master
    WHERE type='table'
    ORDER BY name
    """

    tables = pd.read_sql_query(
        query,
        conn
    )

    conn.close()

    return tables["name"].tolist()


def get_customer_data(customer_id):

    conn = get_connection()

    results = {}

    tables = get_database_tables()

    for table in tables:

        try:

            schema = pd.read_sql_query(
                f"PRAGMA table_info({table})",
                conn
            )

            columns = schema["name"].tolist()

            if "customer_id" in columns:

                query = f"""
                SELECT *
                FROM {table}
                WHERE customer_id = ?
                LIMIT 100
                """

                data = pd.read_sql_query(
                    query,
                    conn,
                    params=(customer_id,)
                )

                results[table] = data

        except Exception as e:

            results[table] = (
                f"Error reading table: {str(e)}"
            )

    conn.close()

    return results


def customer_exists(customer_id):

    conn = get_connection()

    try:

        result = pd.read_sql_query(
            """
            SELECT COUNT(*) AS count
            FROM customers
            WHERE customer_id = ?
            """,
            conn,
            params=(customer_id,)
        )

        exists = result.iloc[0]["count"] > 0

    except Exception:

        exists = False

    conn.close()

    return exists


print("CRM backend loaded successfully")
'''

with open(
    BACKEND_PATH,
    "w",
    encoding="utf-8"
) as f:

    f.write(backend_code)

print("✅ CRM backend created")
print(BACKEND_PATH)
```

```python
# ============================================
#  CRM ANALYTICS TOOL
# ============================================

import sqlite3
import pandas as pd

DB_PATH = "/content/drive/MyDrive/Agentic_AI_CRM/database/crm.db"


def get_crm_summary():

    conn = sqlite3.connect(DB_PATH)

    summary = {}

    summary["total_customers"] = pd.read_sql_query(
        "SELECT COUNT(*) AS count FROM customers",
        conn
    ).iloc[0]["count"]

    summary["total_tickets"] = pd.read_sql_query(
        "SELECT COUNT(*) AS count FROM support_tickets",
        conn
    ).iloc[0]["count"]

    summary["high_priority_tickets"] = pd.read_sql_query(
        """
        SELECT COUNT(*) AS count
        FROM support_tickets
        WHERE LOWER(priority)
        IN ('high', 'urgent', 'critical')
        """,
        conn
    ).iloc[0]["count"]

    summary["open_tickets"] = pd.read_sql_query(
        """
        SELECT COUNT(*) AS count
        FROM support_tickets
        WHERE LOWER(status)
        NOT IN ('closed', 'resolved')
        """,
        conn
    ).iloc[0]["count"]

    summary["negative_sentiment"] = pd.read_sql_query(
        """
        SELECT COUNT(*) AS count
        FROM messages
        WHERE LOWER(overall_sentiment)
        IN ('negative', 'very negative')
        """,
        conn
    ).iloc[0]["count"]

    conn.close()

    return summary


crm_summary = get_crm_summary()

print("📊 CRM SUMMARY")
print("=" * 40)

for key, value in crm_summary.items():
    print(f"{key}: {value:,}")
```

```python
# ============================================
# CRM QUESTION ANSWER TOOL
# ============================================

import sqlite3
import pandas as pd

DB_PATH = "/content/drive/MyDrive/Agentic_AI_CRM/database/crm.db"


def ask_crm(question):

    question_lower = question.lower()

    conn = sqlite3.connect(DB_PATH)

    try:

        # ------------------------------------
        # TOTAL CUSTOMERS
        # ------------------------------------

        if "customer" in question_lower and (
            "how many" in question_lower or
            "total" in question_lower
        ):

            count = pd.read_sql_query(
                "SELECT COUNT(*) AS count FROM customers",
                conn
            ).iloc[0]["count"]

            return f"👥 Total customers: {count:,}"


        # ------------------------------------
        # HIGH PRIORITY TICKETS
        # ------------------------------------

        elif (
            "high priority" in question_lower or
            "critical ticket" in question_lower
        ):

            count = pd.read_sql_query(
                """
                SELECT COUNT(*) AS count
                FROM support_tickets
                WHERE LOWER(priority)
                IN ('high', 'urgent', 'critical')
                """,
                conn
            ).iloc[0]["count"]

            return (
                f"🔴 High-priority tickets: {count:,}"
            )


        # ------------------------------------
        # OPEN TICKETS
        # ------------------------------------

        elif "open ticket" in question_lower:

            count = pd.read_sql_query(
                """
                SELECT COUNT(*) AS count
                FROM support_tickets
                WHERE LOWER(status)
                NOT IN ('closed', 'resolved')
                """,
                conn
            ).iloc[0]["count"]

            return f"📂 Open tickets: {count:,}"


        # ------------------------------------
        # TOP ISSUES
        # ------------------------------------

        elif (
            "top issue" in question_lower or
            "most common issue" in question_lower or
            "biggest problem" in question_lower
        ):

            df = pd.read_sql_query(
                """
                SELECT
                    issue_type,
                    COUNT(*) AS count
                FROM support_tickets
                GROUP BY issue_type
                ORDER BY count DESC
                LIMIT 5
                """,
                conn
            )

            result = "🛠️ Top Support Issues:\n\n"

            for _, row in df.iterrows():

                result += (
                    f"• {row['issue_type']}: "
                    f"{row['count']:,} tickets\n"
                )

            return result


        # ------------------------------------
        # TOP INDUSTRY
        # ------------------------------------

        elif (
            "industry" in question_lower and
            (
                "most" in question_lower or
                "highest" in question_lower
            )
        ):

            df = pd.read_sql_query(
                """
                SELECT
                    industry,
                    COUNT(*) AS count
                FROM support_tickets
                GROUP BY industry
                ORDER BY count DESC
                LIMIT 1
                """,
                conn
            )

            if not df.empty:

                return (
                    f"🏢 Industry with the most tickets: "
                    f"{df.iloc[0]['industry']} "
                    f"({df.iloc[0]['count']:,} tickets)"
                )


        # ------------------------------------
        # NEGATIVE SENTIMENT
        # ------------------------------------

        elif (
            "negative sentiment" in question_lower or
            "negative customer" in question_lower
        ):

            count = pd.read_sql_query(
                """
                SELECT COUNT(*) AS count
                FROM messages
                WHERE LOWER(overall_sentiment)
                IN ('negative', 'very negative')
                """,
                conn
            ).iloc[0]["count"]

            return (
                f"😞 Negative sentiment messages: "
                f"{count:,}"
            )


        # ------------------------------------
        # DEFAULT
        # ------------------------------------

        else:

            return (
                "🤖 I understand your CRM question, "
                "but this question type is not connected yet. "
                "Next we will connect Gemini AI so you can "
                "ask natural language questions."
            )

    except Exception as e:

        return f"❌ Error: {str(e)}"

    finally:

        conn.close()


print("✅ CRM Question Tool created successfully!")
```
